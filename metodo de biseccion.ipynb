## <span style="color:#2980B9">1. Formulación del Problema</span>
El **método de bisección** es un algoritmo de búsqueda de raíces que divide un intervalo a la mitad iterativamente y selecciona el subintervalo que contiene la raíz basándose en el Teorema del Valor Intermedio.

De acuerdo con el **Ejemplo 1 de la página 38 del libro de Burden**, buscamos encontrar la raíz de la siguiente función matemática:
$$ f(x) = x^3 + 4x^2 - 10 = 0 $$

Con este metodo trabajaremos en el intervalo $[1, 2]$ con una tolerancia dada de $10^{-4}$. A continuación, aqui presento la implementación en Python.

```python
import numpy as np
import matplotlib.pyplot as plt
# En tu tarea de métodos numéricos, utilizas numpy para hacer las evaluaciones matemáticas de la bisección de forma eficiente y matplotlib para mostrar la gráfica de la función y cumplir con la presentación requerida.


# 1. Definición de la función para que evaluemos.
# En Python, 'def' se utiliza para declarar una función y 'return' devuelve el valor.
def f(x):
    
    return x**3 + 4*(x**2) - 10

# --- GRAFICA DE LA FUNCION QUE QUEREMOS EVALUAR ---
# np.linspace genera un arreglo de 100 puntos espaciados uniformemente entre 0.5 y 2.5
x_vals = np.linspace(0.5, 2.5, 100) 
y_vals = f(x_vals)

plt.figure(figsize=(8,4))
plt.plot(x_vals, y_vals, label='$f(x) = x^3 + 4x^2 - 10$', color='blue')
plt.axhline(0, color='red', linestyle='--') # Dibuja la línea del eje x
plt.title("Comportamiento de la función en el intervalo [1, 2]")
plt.xlabel("Eje $x$")
plt.ylabel("Eje $y$")
plt.grid(True)
plt.legend()
plt.show()

# --- ALGORITMO 2.1: MÉTODO DE BISECCIÓN (QUE ES EL METODO QUE VAMOS A APLICAR) ---
# Asignación de variables iniciales
a = 1.0       # Límite inferior
b = 2.0       # Límite superior
tol = 1e-4    # Tolerancia (notación científica en Python)
N0 = 50       # Número máximo de iteraciones permitidas

# Paso 1: Inicializar contador y evaluar FA
i = 1
FA = f(a)

# Para una mejor presentación, imprimimos un encabezado. 
# Las llaves {} y los dos puntos :<10 formatean el texto alineándolo a la izquierda.
print(f"{'Iter':<5} | {'a':<10} | {'b':<10} | {'p (Raíz)':<10} | {'f(p)':<10}")
print("-" * 55)

# Paso 2: Ciclo iterativo ('while' ejecuta el bloque mientras la condición sea Verdadera)
while i <= N0:
    # Paso 3: Calcular el punto medio 'p' y evaluarlo
    p = a + (b - a) / 2.0
    FP = f(p)
    
    # Imprimir valores iterativos. La sintaxis .6f obliga a mostrar 6 decimales.
    print(f"{i:<5} | {a:<10.6f} | {b:<10.6f} | {p:<10.6f} | {FP:<10.6f}")
    
    # Paso 4: Criterio de paro. Si f(p) es 0 o la distancia es menor a la tolerancia
    # 'or' es el operador lógico O en Python.
    if FP == 0 or (b - a) / 2.0 < tol:
        raiz_final = p
        break # 'break' fuerza la salida inmediata del ciclo while
        
    # Paso 5: Incrementar el contador (i += 1 es el equivalente a i = i + 1)
    i += 1
    
    # Paso 6: Reasignación de los límites (Condicional if/else)
    if FA * FP > 0:
        a = p
        FA = FP
    else:
        b = p

# Manejo de error si se superan las iteraciones (Paso 7)
if i > N0:
    print(f"El método falló después de {N0} iteraciones")
```

## <span style="color:#27AE60">3. Solución Final y Conclusión</span>

Como se observa en la tabla de iteraciones, el algoritmo convergió exitosamente de acuerdo a los pasos del Algoritmo 2.1 del libro de Burden. El código se detiene cuando el margen de error $(b - a) / 2$ es menor a la tolerancia permitida ($10^{-4}$).
